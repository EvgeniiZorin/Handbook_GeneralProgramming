# AI & ML services in GCP

Levels of customisation and knowledge required:



1. **Cloud AI building blocks**
   - Using models without knowing how they work
   - Examples:
     - Cloud Video Intelligence API
     - Cloud Speech-to-Text
     - Cloud Vision API
     - Cloud Natural Language API
     - Cloud Translation
     - Dialog Flow

2. **AutoML**
    - Suite of ML products that enable developers with limited ML knowledge to train high-quality models specific to their business needs

3. **BigQueryML**
    - Enables using to create and execute ML models by using standard SQL queries
    - Empowers data analysts with limited ML knowledge to develop and use ML models without help of data scientists

4. **AI platform**
    - A set of services build for ML developers, Data scientists and data engineers to build portable ML pipelines that you can run on-premises or Google Cloud
